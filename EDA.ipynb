# 📊 MedStream — Exploratory Data Analysis (EDA)

Welcome to the **Exploratory Data Analysis (EDA)** notebook for **MedStream**!

### What is EDA?
EDA means exploring our data using **charts and statistics** to answer important questions:
- Who is more likely to have diabetes?
- How do blood glucose and HbA1c levels affect diabetes?
- Does age, BMI, or smoking history play a major role?

### What will we do in this notebook?
1. **Load Cleaned Data**: Load `data/processed/diabetes_cleaned.csv`.
2. **Check Summary Statistics**: Look at mean, median, min, and max values.
3. **Target Distribution**: Visualize how many patients have diabetes vs. no diabetes.
4. **Age & BMI Distribution**: See the distribution of age and body mass index.
5. **Blood Tests Comparison**: Compare glucose and HbA1c levels between diabetic and non-diabetic patients.
6. **Hypertension & Heart Disease**: See how high blood pressure and heart disease correlate with diabetes.
7. **Smoking History Analysis**: Compare diabetes rates across different smoking groups.
8. **Correlation Heatmap**: A color-coded matrix showing which features have the strongest relationship with diabetes.

--- 
## Step 1: Import Libraries and Load Cleaned Data
We use **pandas** for data tables, and **matplotlib** & **seaborn** for clean and beautiful plots.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a nice visual style for all plots
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (8, 5)

# Load the cleaned dataset
data_path = "data/processed/diabetes_cleaned.csv"
df = pd.read_csv(data_path)

print("Cleaned dataset loaded successfully!")
print(f"Total records: {len(df):,} patients, {df.shape[1]} columns")
df.head()
```

--- 
## Step 2: Summary Statistics
Let's look at the basic statistics (mean, min, max, standard deviation) for all numeric columns.

```python
# Statistical summary rounded to 2 decimal places
df.describe().round(2)
```

--- 
## Step 3: Target Variable Distribution (Diabetes)
Let's see how many patients have diabetes (`1`) versus no diabetes (`0`).

```python
# Count of patients in each group
counts = df['diabetes'].value_counts()

# Plot bar chart
plt.figure(figsize=(6, 4))
ax = sns.barplot(x=['No Diabetes (0)', 'Has Diabetes (1)'], y=counts.values, palette=['#4C72B0', '#C44E52'])

# Add labels on top of the bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height()):,}', (p.get_x() + 0.3, p.get_height() + 1000))

plt.title("Diabetes Class Distribution", fontsize=14, fontweight='bold')
plt.ylabel("Number of Patients")
plt.show()

# Percentage view
print("Percentage of patients:")
print((df['diabetes'].value_counts(normalize=True) * 100).round(2))
```

--- 
## Step 4: Age Distribution by Diabetes Status
Are older patients more likely to have diabetes? Let's inspect using a Boxplot.

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x='diabetes', y='age', palette=['#55A868', '#C44E52'])

plt.title("Age vs. Diabetes Status", fontsize=14, fontweight='bold')
plt.xticks([0, 1], ['No Diabetes', 'Has Diabetes'])
plt.xlabel("")
plt.ylabel("Age (Years)")
plt.show()
```

--- 
## Step 5: Blood Glucose & HbA1c Comparison
`blood_glucose_level` and `HbA1c_level` are the most critical medical indicators for diabetes.
Let's see the average values for each group.

```python
# Compare average blood glucose and HbA1c levels
avg_metrics = df.groupby('diabetes')[['blood_glucose_level', 'HbA1c_level', 'bmi']].mean().round(2)
print("=== Average Values by Diabetes Status ===")
avg_metrics.index = ['No Diabetes', 'Has Diabetes']
print(avg_metrics)

# Visual comparison: Blood Glucose Level
plt.figure(figsize=(8, 4))
sns.kdeplot(data=df, x='blood_glucose_level', hue='diabetes', common_norm=False, fill=True, palette=['#4C72B0', '#C44E52'])
plt.title("Blood Glucose Level Density (Diabetic vs Non-Diabetic)", fontsize=13, fontweight='bold')
plt.xlabel("Blood Glucose Level (mg/dL)")
plt.show()
```

--- 
## Step 6: Hypertension and Heart Disease Impact
Let's see what percentage of patients with hypertension (high blood pressure) or heart disease have diabetes.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 1. Hypertension vs Diabetes Rate
sns.barplot(data=df, x='hypertension', y='diabetes', ax=axes[0], palette='Blues_d', errorbar=None)
axes[0].set_title("Diabetes Rate by Hypertension", fontweight='bold')
axes[0].set_xticklabels(['No Hypertension', 'Has Hypertension'])
axes[0].set_ylabel("Diabetes Proportion")

# 2. Heart Disease vs Diabetes Rate
sns.barplot(data=df, x='heart_disease', y='diabetes', ax=axes[1], palette='Reds_d', errorbar=None)
axes[1].set_title("Diabetes Rate by Heart Disease", fontweight='bold')
axes[1].set_xticklabels(['No Heart Disease', 'Has Heart Disease'])
axes[1].set_ylabel("Diabetes Proportion")

plt.tight_layout()
plt.show()
```

--- 
## Step 7: Smoking History Comparison
Does smoking history correlate with diabetes diagnosis?

```python
plt.figure(figsize=(8, 4))
order = ['never', 'former', 'current', 'No Info']
sns.barplot(data=df, x='smoking_history', y='diabetes', order=order, palette='crest', errorbar=None)

plt.title("Diabetes Proportion by Smoking History", fontsize=13, fontweight='bold')
plt.xlabel("Smoking History Category")
plt.ylabel("Diabetes Proportion")
plt.show()
```

--- 
## Step 8: Correlation Heatmap
A heatmap shows how strongly numeric features relate to each other and to **diabetes**:
- Closer to **+1.0**: Strong positive relationship.
- Closer to **0.0**: Little or no relationship.

```python
# Select only numeric columns for correlation
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title("Feature Correlation Matrix", fontsize=14, fontweight='bold')
plt.show()

# Direct correlation with Diabetes target
print("\n=== Correlation with Diabetes Target ===")
print(corr_matrix['diabetes'].sort_values(ascending=False).round(3))
```
