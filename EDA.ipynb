# 🩺 تحليل بيانات مرض السكري (EDA بأبسط طريقة ممكنة)

أهلاً بيك يا صديقي! 👋  
الملف ده معمول عشان نفهم الداتا بتاعتنا من غير أي لف ودوران ولا أكواد معقدة.  
كل خطوة هنا بتوريك **الرسمة + جدول صغير فيه المتوسط (Mean) والوسيط (Median)** عشان الأرقام تكون واضحة في عينك تماماً! 🚀

--- 
## الخطوة 1: نفتح الداتا بتاعتنا
**الشرح بالمصري:**  
هنا بننادي مكتبة `pandas` المسؤولة عن الجداول، ونقرأ ملف الداتا النظيفة اللي جهزناها عشان نبدأ نشتغل عليها.

```python
import pandas as pd

# بنقرأ الداتا
df = pd.read_csv('data/processed/diabetes_cleaned.csv')

# بنشوف أول 5 صفوف علشان نتطمن إن الداتا تمام
df.head()
```

--- 
## الخطوة 2: نبذة سريعة بالأرقام عن المرضى
**الشرح بالمصري:**  
كود `df.describe()` ده عامل زي التقرير الشامل، بيطلعلك في ثانية واحدة:
- متوسط أعمار المرضى كام سنة.
- متوسط السكر التراكمي ونسبة الجلوكوز في الدم.
- أقل وأعلى وزن وسن موجودين في الداتا.

```python
# جدول أرقام وإحصائيات سريعة ومختصرة
df.describe().round(1)
```

--- 
## الخطوة 3: كام واحد عنده سكر وكام واحد سليم؟ (الهدف بتاعنا)
**الشرح بالمصري:**  
هنا بنعد الناس اللي عندها سكر (`1`) والناس السليمة (`0`)، وبنعرض الأعداد والنسبة المئوية مع الرسمة.

```python
# بنعد الحالات ونرسمها: 0 يعني سليم و 1 يعني عنده سكر
print("=== أعداد الحالات والنسب ===")
print(df['diabetes'].value_counts())
df['diabetes'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'], title='0 = Non-Diabetic, 1 = Diabetic')
```

--- 
## الخطوة 4: أعمار المرضى عاملة إزاي؟
**الشرح بالمصري:**  
بنعرف متوسط ووسيط عمر الناس في الداتا كلها، وبنرسم توزيع الأعمار.

```python
# متوسط ووسيط العمر لكل الناس في الداتا
print(f"متوسط العمر (Mean): {df['age'].mean():.1f} سنة")
print(f"وسيط العمر (Median): {df['age'].median():.1f} سنة")

# رسم بياني لتوزيع الأعمار
df['age'].plot(kind='hist', bins=10, color='lightgreen', title='Distribution of Patient Age')
```

--- 
## الخطوة 5: هل السن ليه علاقة بالإصابة بالسكر؟
**الشرح بالمصري:**  
هنا بنحسب **متوسط (Mean)** و **وسيط (Median)** سن الشخص السليم مقارنة بالشخص مريض السكر:
- السليم (0): متوسط سنه 40 سنة.
- مريض السكر (1): متوسط سنه 61 سنة (فرق 21 سنة!).

```python
# جدول المتوسط والوسيط للأعمار حسب الإصابة
print("=== إحصائيات السن لكل مجموعة ===")
print(df.groupby('diabetes')['age'].agg(['mean', 'median']).round(2))

# مقارنة سن الأصحاء (0) بسن المصابين بالسكر (1) في رسمة الصندوق
df.boxplot(column='age', by='diabetes')
```

--- 
## الخطوة 6: مستوى السكر في الدم (`blood_glucose_level`)
**الشرح بالمصري:**  
بنعرض متوسط ووسيط سكر الدم:
- السليم (0): متوسط سكر دمه 132.
- مريض السكر (1): متوسط سكر دمه 194 (مرتفع جداً!).

```python
# جدول المتوسط والوسيط لسكر الدم
print("=== إحصائيات سكر الدم لكل مجموعة ===")
print(df.groupby('diabetes')['blood_glucose_level'].agg(['mean', 'median']).round(1))

# رسمة الصندوق لمقارنة سكر الدم
df.boxplot(column='blood_glucose_level', by='diabetes')
```

--- 
## الخطوة 7: السكر التراكمي (`HbA1c_level`)
**الشرح بالمصري:**  
بنعرض متوسط ووسيط السكر التراكمي:
- السليم (0): متوسط التراكمي عنده 5.5 (في المعدل الطبيعي).
- مريض السكر (1): متوسط التراكمي عنده 6.7 (في مرحلة مرض السكري المؤكدة).

```python
# جدول المتوسط والوسيط للسكر التراكمي
print("=== إحصائيات السكر التراكمي لكل مجموعة ===")
print(df.groupby('diabetes')['HbA1c_level'].agg(['mean', 'median']).round(1))

# رسمة الصندوق لمقارنة السكر التراكمي
df.boxplot(column='HbA1c_level', by='diabetes')
```

--- 
## الخطوة 8: مؤشر كتلة الجسم والوزن (`bmi`)
**الشرح بالمصري:**  
بنشوف هل الوزن والـ BMI بيأثروا في السكر:
- السليم (0): متوسط الـ BMI حوالي 26.9.
- مريض السكر (1): متوسط الـ BMI حوالي 32.0 (مرحلة السمنة).

```python
# جدول المتوسط والوسيط للـ BMI
print("=== إحصائيات الـ BMI لكل مجموعة ===")
print(df.groupby('diabetes')['bmi'].agg(['mean', 'median']).round(1))

# رسمة الصندوق للوزن والـ BMI
df.boxplot(column='bmi', by='diabetes')
```

--- 
## الخطوة 9: إيه أكتر حاجات مرتبطة بمرض السكر؟ (الارتباط)
**الشرح بالمصري:**  
الكود ده بيرتبلك العوامل من حيث قوة ارتباطها بمرض السكر بالأرقام:
الأول بلا منازع: سكر الدم والتراكمي، ثم السن والـ BMI والضغط.

```python
# ترتيب العوامل الأكثر ارتباطاً بالسكر
df.corr(numeric_only=True)['diabetes'].sort_values(ascending=False).round(3)
```
