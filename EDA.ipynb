### يعني إيه EDA أصلاً؟
الـ **EDA** يعني إننا نبص على البيانات ونستكشفها بالبلدي كده:  
- كام واحد عنده سكر وكام واحد سليم؟  
- مين بيجيله سكر أكتر: الكبار ولا الصغيرين؟  
- هل الضغط والقلب ليهم علاقة بالسكر؟  
- إيه أكتر تحاليل بتبين المرض على طول؟

--- 
## الخطوة 1: نفتح الداتا بتاعتنا
**الشرح بالمصري:**  
هنا بننادي مكتبة `pandas` المسؤولة عن الجداول، ونقرأ ملف الداتا النظيفة اللي جهزناها عشان نبدأ نشتغل عليها.

```python
import pandas as pd

# بنقرأ الداتا
df = pd.read_csv('data/processed/diabetes_cleaned.csv')

# بنشوف أول 5 صفوف علشان نتطمن إن الداتا تمام
df.head()
```

--- 
## الخطوة 2: نبذة سريعة بالأرقام عن المرضى
**الشرح بالمصري:**  
كود `df.describe()` ده عامل زي التقرير الشامل، بيطلعلك في ثانية واحدة:
- متوسط أعمار المرضى كام سنة.
- متوسط السكر التراكمي ونسبة الجلوكوز في الدم.
- أقل وأعلى وزن وسن موجودين في الداتا.

```python
# جدول أرقام وإحصائيات سريعة ومختصرة
df.describe().round(1)
```

--- 
## الخطوة 3: كام واحد عنده سكر وكام واحد سليم؟ (الهدف بتاعنا)
**الشرح بالمصري:**  
هنا بنعد الناس اللي عندها سكر (`1`) والناس السليمة (`0`)، وبنرسمها في شكل عواميد ملونة وسهلة عشان توضح الفرق.

```python
# بنعد ونرسم في سطر واحد: 0 يعني سليم و 1 يعني عنده سكر
df['diabetes'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'], title='0 = Non-Diabetic, 1 = Diabetic')
```

--- 
## الخطوة 4: أعمار المرضى عاملة إزاي؟
**الشرح بالمصري:**  
عايزين نعرف معظم الناس اللي في الداتا سنهم قد إيه؟ هل شباب ولا كبار سن؟  
رسمة الهيستوجرام (`hist`) بتوريك الفئات العمرية الأكثر تكراراً في الداتا.

```python
# رسم بياني لتوزيع الأعمار
df['age'].plot(kind='hist', bins=10, color='lightgreen', title='Distribution of Patient Age')
```

--- 
## الخطوة 5: هل السن ليه علاقة بالإصابة بالسكر؟
**الشرح بالمصري:**  
هنا بنقارن أعمار المصابين بأعمار الأصحاء عن طريق الـ Boxplot.  
حتلاحظ علطول إن الصندوق الخاص بمرضى السكري (`1`) طالع لفوق عند الأعمار الأكبر، وده بياكد إن كبار السن هما الأكثر عرضة للمرض.

```python
# مقارنة سن الأصحاء (0) بسن المصابين بالسكر (1)
df.boxplot(column='age', by='diabetes')
```

--- 
## الخطوة 6: مستوى السكر في الدم (`blood_glucose_level`)
**الشرح بالمصري:**  
ده التحليل الأهم طبياً! بنقارن نسبة السكر في دم الناس السليمة بالناس التعبانة.  
حيبان جداً إن مرضى السكر أرقامهم معدية الـ 200 بسهولة عكس الأصحاء.

```python
# مقارنة نسبة السكر في الدم بين السليم والمريض
df.boxplot(column='blood_glucose_level', by='diabetes')
```

--- 
## الخطوة 7: السكر التراكمي (`HbA1c_level`)
**الشرح بالمصري:**  
التراكمي ده التحليل اللي بيقيس السكر لـ 3 شهور فاتوا:  
الشخص الطبيعي بيكون أقل من 5.7، أما مريض السكر بيكون فوق 6.5.  
الرسمة دي هتوريك الفرق الواضح بين الفئتين.

```python
# مقارنة السكر التراكمي بين السليم والمريض
df.boxplot(column='HbA1c_level', by='diabetes')
```

--- 
## الخطوة 8: التدخين والسكر
**الشرح بالمصري:**  
عايزين نعرف توزيع الناس حسب عادة التدخين:  
مين أكتر؟ اللي مابيدخنوش خالص، ولا اللي دخنوا زمان وبطلوا، ولا المدخنين حالياً؟

```python
# رسمة فئات التدخين
df['smoking_history'].value_counts().plot(kind='bar', color='orange', title='Smoking History Counts')
```

--- 
## الخطوة 9: إيه أكتر حاجات مرتبطة بمرض السكر؟ (الارتباط)
**الشرح بالمصري:**  
الكود البسيط ده بيحسبلك علاقة كل حاجة بالسكر بنمرة ورقم:  
- لو الرقم كبير وقريب من 1 يبقى بيأثر جداً في وجود السكر.  
- حتلاحظ إن أعلى حاجتين هما: `blood_glucose_level` (سكر الدم) و `HbA1c_level` (التراكمي)، ويليهم السن والضغط.

```python
# بنحسب ونرتب أكتر العوامل المؤثرة في السكر
df.corr(numeric_only=True)['diabetes'].sort_values(ascending=False)
```
