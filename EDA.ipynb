# MedStream - Exploratory Data Analysis

Analysis of patient features against diabetes outcome on cleaned records (`diabetes_cleaned.csv`).

### 1. Load Data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data/processed/diabetes_cleaned.csv')
print(f"Loaded {len(df):,} rows")
df.head()
```

### 2. Summary Statistics

```python
df.describe().round(2)
```

### 3. Diabetes Class Distribution

```python
print(df['diabetes'].value_counts())
df['diabetes'].value_counts().plot(kind='bar', color=['#4C72B0', '#C44E52'], title='0: Non-Diabetic | 1: Diabetic')
plt.ylabel('Count')
plt.show()
```

### 4. Age Distribution

```python
print(f"Mean age: {df['age'].mean():.1f} | Median age: {df['age'].median():.1f}")
df['age'].plot(kind='hist', bins=15, edgecolor='black', title='Age Distribution')
plt.xlabel('Age (years)')
plt.show()
```

### 5. Age by Diabetes Status

```python
print(df.groupby('diabetes')['age'].agg(['mean', 'median', 'std']).round(1))
df.boxplot(column='age', by='diabetes')
plt.title('Age by Diabetes Status')
plt.suptitle('')
plt.xlabel('Diabetes (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()
```

### 6. Blood Glucose Level Comparison

```python
print(df.groupby('diabetes')['blood_glucose_level'].agg(['mean', 'median']).round(1))
df.boxplot(column='blood_glucose_level', by='diabetes')
plt.title('Blood Glucose Level by Diabetes')
plt.suptitle('')
plt.xlabel('Diabetes')
plt.ylabel('Blood Glucose (mg/dL)')
plt.show()
```

### 7. HbA1c Level Comparison

```python
print(df.groupby('diabetes')['HbA1c_level'].agg(['mean', 'median']).round(2))
df.boxplot(column='HbA1c_level', by='diabetes')
plt.title('HbA1c Level by Diabetes')
plt.suptitle('')
plt.xlabel('Diabetes')
plt.ylabel('HbA1c')
plt.show()
```

### 8. BMI Comparison

```python
print(df.groupby('diabetes')['bmi'].agg(['mean', 'median']).round(1))
df.boxplot(column='bmi', by='diabetes')
plt.title('BMI by Diabetes')
plt.suptitle('')
plt.xlabel('Diabetes')
plt.ylabel('BMI')
plt.show()
```

### 9. Smoking History Breakdown

```python
print(df['smoking_history'].value_counts())
df['smoking_history'].value_counts().plot(kind='bar', color='#55A868', title='Smoking Status Counts')
plt.ylabel('Patients')
plt.show()
```

### 10. Feature Correlations

```python
corr = df.corr(numeric_only=True)['diabetes'].sort_values(ascending=False).round(3)
print("Correlation with diabetes target:")
print(corr)
```
